# **2.Understanding JSON**

````python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/todos/1")

data = response.json()

print(data)
print(type(data))
print(data["title"])
````

## **Key Takeaway**
JSON is the standard format applications use to exchange structured information.

# **3.POST Request**

URL:
https://postman-echo.com/post


````python
import requests

payload = {"name": "Gabriel",
           "company": "PROITBRIDGE",
           "role": "AI Engineer"}

response = requests.post("https://postman-echo.com/post", json=payload, timeout=30)

print("Status Code:", response.status_code)
print(response.json())
````

### **Key Takeaway**

- GET retrieves information.
- POST sends structured information.
- Modern AI APIs internally use the same communication principles.

# **5. Understanding the Engineering Problem**

````python
from openai import OpenAI

client = OpenAI(api_key="Paste_YOUR_API_KEY")

question = input("Ask something: ")

response = client.chat.completions.create(
    model="gpt-5.6",
    messages=[
        {
            "role":"user",
            "content":question
        }
    ]
)

print(response.choices[0].message.content)
````

### ***The model forgets***.

### Why?

Every API request is **independent (stateless)**.

The model only knows the information sent in the **current request**.

If our application asks multiple questions, it must also send:

- Context
- Conversation History
- Previous Messages

This is why **Memory** becomes an engineering requirement.
